### Regular Expression Example and Coding Simple Tokenizer with Regular Expression
- Funtions to Wrap Text and Wrap Lists without List to String Conversion

````python
# @title Functions and Function Classes
import textwrap

# Function to Wrap text by converting list to string
def printWT(label, data, width=80): # Wrap Text into Frame
    """Converts the data list/object to a string and wraps the text smoothly across lines."""
    print(label)
    wrapped_text = textwrap.fill(str(data), width=width)
    print(wrapped_text)
    print("-" * width)

# Function to Wrap Long Lists without losing the list structure representation
def printWL(label, items, width=80): # Function to Wrap Long List Frame 
    """Wraps long lists by appending items and managing line breaks without exceeding the width."""
    print(label)

    current_line = "["
    for item in items:
        formatted_item = repr(item) + ", "

        # Check if adding this item exceeds the screen width
        if len(current_line) + len(formatted_item) > width - 1:
            print(current_line)
            current_line = " " + formatted_item  # Start next line indented
        else:
            current_line += formatted_item

    # Print remaining items and close the bracket
    if current_line:
        print(current_line.rstrip(", ") + "]")
    print("-" * width)

# Function to wrap and format as a Python Dictionary { key: value }
def printWD(label, dictionary, width=80, max_items=None): # Function to Wrap Dictionary 
    """Wraps and formats a dictionary as a string, keeping lines within the maximum width.
    Optionally limits the print output to the first max_items if specified.
    """
    print(label)

    current_line = "{"
    items = list(dictionary.items())
    if max_items is not len(items) and max_items is not None:
        items = items[:max_items]

    for i, (key, value) in enumerate(items):
        # Format key and value with Python repr (keeps strings quoted)
        entry = f"{repr(key)}: {repr(value)}"
        if i < len(items) - 1:
            entry += ", "

        # Check if adding this entry exceeds line width
        if len(current_line) + len(entry) > width - 1:
            print(current_line)
            current_line = "  " + entry  # Indent continuation line
        else:
            current_line += entry

    if current_line:
        closing = "}" if max_items is None or len(dictionary) <= max_items else "...}"
        print(current_line.rstrip(", ") + closing)
    print("-" * width)

# Function to display a dictionary in a clean, column-aligned table
def printDT(label, dictionary, max_key_len=20, max_val_len=50):
    """Prints a dictionary as an aligned key-value table, truncating long items."""
    print(label)
    print(f"{'KEY':<{max_key_len}} | {'VALUE':<{max_val_len}}")
    print("-" * (max_key_len + max_val_len + 3))

    for key, value in dictionary.items():
        k_str = str(key)[:max_key_len - 3] + "..." if len(str(key)) > max_key_len else str(key)
        v_str = str(value)[:max_val_len - 3] + "..." if len(str(value)) > max_val_len else str(value)
        print(f"{k_str:<{max_key_len}} | {v_str:<{max_val_len}}")
````

````python
# @title Understaning Regular Expression and Coding Simple Tokenizer
import re


example_text="Hello, world. This, is a test."

# 1st split for whitespaces(\s)
split0= re.split(r'(\s)', example_text)
printWL("1st split for emply spaces: ",split0)

# 2nd split on whitespaces(\s), commas(,) and periods(,)
split1= re.split(r'([,.]|\s)', example_text)
printWL("2nd split on whitespaces, \ncommas(,) and periods(,): ", split1)

#3rd Coding Simple Tokenizer with Regular Expression
split2= re.split(r'([,.:;?_!"()\']|--|\s)',example_text)
printWL("3rd split with simple tokenizer splits on white spaces, and all punchuations: ", split2[:200])

# Generating Tokenized Text
result= [word.strip() for word in split2 if word.strip()]
printWL("Tokenized Text: ",result[:200])
````

## Converting token into token IDs

````python
all_words=sorted(set(preprocessed))
vocab_size=len(all_words)
printWT("Sorted all words: ",all_words[:50])
print(vocab_size)
````

````python
vocab={token:integer for integer, token in enumerate(all_words)}
printWD("Vocab: ",vocab, max_items=50)
````

````python
# @title Implementing a simple text tokenizer

class SimpleTokenizerV1:

    def __init__(self, vocab):
      self.str_to_int = vocab
      # self.vocab_size = len(vocab)
      self.int_to_str = {v: k for k, v in self.str_to_int.items()}



    def encode(self, vocab):
      preprocessed = re.split(r'([,.:;?_!"()\']|--|\s)', vocab)
      preprocessed = [
          word.strip() for word in preprocessed if word.strip()
          ]
      ids= [self.str_to_int[word] for word in preprocessed]
      return ids

    def decode(self, ids):
      text = " ".join([self.int_to_str[i] for i in ids])
      text = re.sub(r'\s+([,.?!"()\'])', r'\1', text)
      return text
````

````python
tokenizer = SimpleTokenizerV1(vocab)

text = """"It's the last he painted, you know,"
           Mrs. Gisburn said with pardonable pride."""
ids = tokenizer.encode(text)
print(ids)
````

````python
tokenizer.decode(ids)
````

````python
tokenizer.decode(tokenizer.encode(text))
````
