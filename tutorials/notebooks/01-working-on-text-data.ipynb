````
# @title Functions and Function Classes
import textwrap

# Function to Wrap text by converting list to string
def printWT(label, data, width=80):
    """Converts the data list/object to a string and wraps the text smoothly across lines."""
    print(label)
    wrapped_text = textwrap.fill(str(data), width=width)
    print(wrapped_text)
    print("-" * width)

# Function to Wrap Long Lists without losing the list structure representation
def printWL(label, items, width=80):
    """Wraps long lists by appending items and managing line breaks without exceeding the width."""
    print(label)
    if not items:
        print("[]")
        print("-" * width)
        return

    lines = []
    current_line = "["

    for i, item in enumerate(items):
        formatted_item = repr(item)
        is_last = (i == len(items) - 1)

        # Prepare what we want to append
        suffix = "" if is_last else ", "
        item_str = formatted_item + suffix

        # If adding this item to the current line exceeds width (excluding the final closed bracket)
        # and the current line isn't just the starting bracket
        if len(current_line) + len(item_str) > (width - 1 if is_last else width) and current_line != "[":
            lines.append(current_line)
            current_line = " " + item_str
        else:
            current_line += item_str

    if current_line:
        current_line += "]"
        lines.append(current_line)

    for line in lines:
        print(line)
    print("-" * width)

# Function to wrap and format as a Python Dictionary { key: value }
def printWD(label, dictionary, width=80, max_items=None):
    """Wraps and formats a dictionary as a string, keeping lines within the maximum width.
    Optionally limits the print output to the first max_items if specified.
    """
    print(label)

    current_line = "{"
    items = list(dictionary.items())
    if max_items is not len(items) and max_items is not None:
        items = items[:max_items]

    for i, (key, value) in enumerate(items):
        entry = f"{repr(key)}: {repr(value)}"
        if i < len(items) - 1:
            entry += ", "

        if len(current_line) + len(entry) > width - 1:
            print(current_line)
            current_line = "  " + entry
        else:
            current_line += entry

    if current_line:
        closing = "}" if max_items is None or len(dictionary) <= max_items else "...}"
        print(current_line.rstrip(", ") + closing)
    print("-" * width)

# Function to display a dictionary in a clean, column-aligned table
def printDT(label, dictionary, max_key_len=20, max_val_len=50):
    """Prints a dictionary as an aligned key-value table, truncating long items."""
    print(label)
    print(f"{'KEY':<{max_key_len}} | {'VALUE':<{max_val_len}}")
    print("-" * (max_key_len + max_val_len + 3))

    for key, value in dictionary.items():
        k_str = str(key)[:max_key_len - 3] + "..." if len(str(key)) > max_key_len else str(key)
        v_str = str(value)[:max_val_len - 3] + "..." if len(str(value)) > max_val_len else str(value)
        print(f"{k_str:<{max_key_len}} | {v_str:<{max_val_len}}")
````

````
# @title Understaning Regular Expression and Coding Simple Tokenizer
import re


example_text="Hello, world. This, is a test."

# 1st split for whitespaces(\s)
split0= re.split(r'(\s)', example_text)
printWL("1st split for emply spaces: ",split0)

# 2nd split on whitespaces(\s), commas(,) and periods(,)
split1= re.split(r'([,.]|\s)', example_text)
printWL("2nd split on whitespaces, \ncommas(,) and periods(,): ", split1)

#3rd Coding Simple Tokenizer with Regular Expression
split2= re.split(r'([,.:;?_!"()\']|--|\s)',example_text)
printWL("3rd split with simple tokenizer splits on white spaces, and all punchuations: ", split2[:200])

# Generating Tokenized Text
result= [word.strip() for word in split2 if word.strip()]
printWL("Tokenized Text: ",result[:200])
````

````
# @title Implementing Simple Tokenizers

class SimpleTokenizerV1:

    def __init__(self, vocab):
      self.str_to_int = vocab
      self.int_to_str = {v: k for k, v in self.str_to_int.items()}



    def encode(self, vocab):
      preprocessed = re.split(r'([,.:;?_!"()\']|--|\s)', vocab)
      preprocessed = [
          word.strip() for word in preprocessed if word.strip()
          ]
      ids= [self.str_to_int[word] for word in preprocessed]
      return ids

    def decode(self, ids):
      text = " ".join([self.int_to_str[i] for i in ids])
      text = re.sub(r'\s+([,.?!"()\'])', r'\1', text)
      return text

tokenizerV1 = SimpleTokenizerV1(vocab)



# @title Implementing a simple text tokenizer
"""A Simple TokenizerV2 which handles "unknown" words
"""

class SimpleTokenizerV2:


    def __init__(self, upd_vocab):
      self.str_to_int = upd_vocab
      self.int_to_str = {v: k for k, v in self.str_to_int.items()}

    def encode(self, vocab):
      # We update the regex to capture the special tokens whole: (<\|endoftext\|>|<\|eos\|>|<\|unk\|>)
      preprocessed = re.split(r'(<\|endoftext\|>|<\|eos\|>|<\|unk\|>|[,.:;?_!"()\']|--|\s)', vocab)
      preprocessed = [
          item.strip() for item in preprocessed if item.strip()
          ]
      preprocessed = [item if item in self.str_to_int else "<|unk|>" for item in preprocessed]
      ids = [self.str_to_int[item] for item in preprocessed]
      return ids

    def decode(self, ids):
      text = " ".join([self.int_to_str[i] for i in ids])
      # Replace spaces before the specified punctuations
      text = re.sub(r'\s+([,.:;?!"()\'])', r'\1', text)
      return text
````

````
# @title Coding Dataset Class and DataLoader

import torch
from torch.utils.data import Dataset, DataLoader

class GPTDatasetV1(Dataset):
  """
  A custom PyTorch Dataset that prepares tokenized text into input-target chunk pairs
  using a sliding window approach for training Autoregressive Language Models (like GPT).
  """

  def __init__(self, txt, tokenizer, max_length, stride):
    """
    Initializes the dataset by tokenizing the input text and sliding a window over it
    to create input chunks (x) and target chunks (y) shifted by one token.

    Args:
        txt (str): The raw input text document.
        tokenizer (tiktoken.Encoding): The Byte-Pair-Encoding tokenizer used to encode text.
        max_length (int): The sequence length (context size) of each training sample.
        stride (int): The distance the sliding window shifts at each step.
    """
    # Two empty lists are initialized to store the chunks of input data and their corresponding shifted targets.
    self.input_ids = []
    self.target_ids = []

    # Tokenizing the entire input text into a flat list of token IDs.
    # The parameter allowed_special={"<|endoftext|"} allows processing specific special end-of-text markers if present.
    token_ids = tokenizer.encode(txt, allowed_special={"<|endoftext|"})

    # Safety check: ensure we have enough tokens to form at least one full input (max_length) and target (shifted by 1) slice.
    assert len(token_ids) > max_length, "Number of tokenized inputs must at least be equal to max_length+1"

    # Slide a window across the tokens to generate overlapping chunk pairs.
    for i in range(0, len(token_ids) - max_length, stride):
      input_chunk = token_ids[i : i + max_length]
      # Target chunk is the exact same window, but shifted forward by 1 token to predict the next word
      target_chunk = token_ids[i + 1 : i + max_length + 1]

      # Store both chunks as PyTorch Tensors
      self.input_ids.append(torch.tensor(input_chunk))
      self.target_ids.append(torch.tensor(target_chunk))

  def __len__(self):
    """
    Returns the total number of sliding window samples generated.
    """
    return len(self.input_ids)

  def __getitem__(self, idx):
    """
    Retrieves a single training pair (input_chunk, target_chunk) at the specified index.
    """
    return self.input_ids[idx], self.target_ids[idx]
````

````
# @title defining Build_DataLoaderV1 function to build Pytorch based Data Loader
def Build_DataLoaderV1(txt,
                       batch_size=4,
                       max_length=256,
                       stride=128,
                       shuffle=True,
                       drop_last=True,
                       num_workers=0
                       ):
  """
  Constructs and returns a PyTorch DataLoader for tokenized text inputs using a sliding window.

  Args:
      txt (str): The raw text sequence to tokenize and load.
      batch_size (int): Number of text sequences grouped together in each batch. Defaults to 4.
      max_length (int): Maximum sequence length (context window) for inputs and targets. Defaults to 256.
      stride (int): Step size to slide the context window across the tokenized text. Defaults to 128.
      shuffle (bool): If True, shuffles the dataset batches before each training epoch. Defaults to True.
      drop_last (bool): If True, discards the last batch if its size is smaller than batch_size. Defaults to True.
      num_workers (int): Number of subprocesses to use for data loading. Defaults to 0.

  Returns:
      DataLoader: A PyTorch DataLoader iterator yielding (input_batch, target_batch) token ID pairs.
  """
  # Load the standard GPT-2 byte-pair encoding (BPE) tokenizer
  tokenizer = tiktoken.get_encoding("gpt2")

  # Initialize the dataset which handles tokenization and sliding window division
  dataset = GPTDatasetV1(txt, tokenizer, max_length, stride)

  # Package the dataset into a PyTorch DataLoader configuration
  dataloader = DataLoader(
      dataset,
      batch_size=batch_size,
      shuffle=shuffle,
      drop_last=drop_last,
      num_workers=num_workers
      )

  return dataloader
````

````
def printBD(batch_name, batch, tokenizer=None, width=50):
    """
    An upgraded, highly reusable helper function to cleanly print, format,
    and optionally decode inputs and targets from PyTorch DataLoader batches.

    Args:
        batch_name (str): Title or label for the current batch printout.
        batch (tuple/list): A pair of tensors (inputs, targets).
        tokenizer (optional): A tokenizer object (e.g. BPE_tokenizer) with a .decode() method
                             to translate token IDs back to plain text.
        width (int): Formatting width of decorative separator lines.
    """
    # Format header nicely aligned
    header = f" {batch_name} ".center(width, "=")
    print(header)

    inputs, targets = batch

    # Display Inputs metadata and content
    print(f"Inputs (Shape: {list(inputs.shape)}):")
    print(inputs)
    if tokenizer is not None:
        print("Decoded Inputs:")
        for i, seq in enumerate(inputs):
            # Handle batched data or single sequences
            seq_list = seq.tolist() if hasattr(seq, 'tolist') else list(seq)
            decoded = tokenizer.decode(seq_list)
            print(f"  Seq {i + 1}: {decoded}")

    print("-" * width)

    # Display Targets metadata and content
    print(f"Targets (Shape: {list(targets.shape)}):")
    print(targets)
    if tokenizer is not None:
        print("Decoded Targets:")
        for i, seq in enumerate(targets):
            seq_list = seq.tolist() if hasattr(seq, 'tolist') else list(seq)
            decoded = tokenizer.decode(seq_list)
            print(f"  Seq {i + 1}: {decoded}")

    print("=" * width + "\n")
````
